```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.datasets import load_iris
import pandas as pd
iris=load_iris()
iris_data=pd.DataFrame(iris.data,columns=iris.feature_names)
iris_data.head(10)
```

```python
from sklearn.neighbors import NearestNeighbors
from sklearn.datasets import make_blobs
x,_=make_blobs(n_samples=len(iris_data.index),centers=75,cluster_std=0.60, random_state=42)
k = 5

# 3. Compute distances to the k nearest neighbors
nbrs = NearestNeighbors(n_neighbors=k).fit(x)
distances, _ = nbrs.kneighbors(x)

# 4. Extract distance to the k-th neighbor and sort in ascending order
k_distances = np.sort(distances[:, k - 1])

# 5. Plot the k-distance graph
plt.figure(figsize=(8, 5))
plt.plot(k_distances, color="#1f77b4", linewidth=2)
plt.title(f"{k}-Distance Graph for DBSCAN `eps` Selection", fontsize=12)
plt.xlabel("Points Sorted by Distance", fontsize=11)
plt.ylabel(f"{k}-th Nearest Neighbor Distance", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
dbscan=DBSCAN(eps=0.8,min_samples=5)
labels=dbscan.fit_predict(iris_data)
print("Cluster Labels:")
print(labels)
n_cluster=len(set(labels))-(1 if -1 in labels else 0)
print("Number of Clusters:",n_cluster)
n_noise=list(labels).count(-1)
print("Number of Noise Points:",n_noise)
print("Core point Indices:")
print(dbscan.core_sample_indices_)
```

```python
X=iris_data[['petal length (cm)','petal width (cm)']]
plt.figure(figsize=(8,5))
plt.scatter(X['petal length (cm)'],X['petal width (cm)'],c=labels)
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.title('DBSCAN Clustering')
plt.show()
```
