```python
import numpy as np
from scipy import stats
mu_0,sigma,n,smean,alpha=3.25,1.61,900,3.4,0.05
SE=sigma/np.sqrt(n)
z_stat=(smean-mu_0)/SE
p_value=2*stats.norm.sf(abs(z_stat))
print(f"P value:{p_value:.4f}")
if p_value<alpha:
    print("Reject null hypothesis")
else:
    print("Fail to reject null hypothesis")
```

```python
n1,n2,m1,m2,s1,s2=100,64,18.2,19,4,5
SE=np.sqrt((s1**2/n1)+(s2**2/n2))
z_stat=(m1-m2)/SE
print(z_stat)
p_value=2*stats.norm.sf(abs(z_stat))
print(f"P value:{p_value:.4f}")
alpha=0.05
if p_value<alpha:
    print("Reject null hypothesis")
else:
    print("Fail to reject null hypothesis")
```

```python
import numpy as np
from scipy import stats
m,mu,s,n=0.742,0.7,0.04,10
SE=s/np.sqrt(n-1)
t_cal=(m-mu)/SE
df=n-1
p_value=stats.t.sf(abs(t_cal),df)
print(f"P value:{p_value:.4f}")
if p_value<0.025:
    print("Reject null hypothesis")
else:
    print("Fail to reject null hypothesis")
```

```python
sample=[5,2,8,-1,3,0,-2,1,5,0,4,6]
n=len(sample)
m,s=np.mean(sample),np.std(sample)
mu,df=0,n-1
SE=s/np.sqrt(n-1)
t_cal=abs((m-mu)/SE)
p_value=stats.t.sf(abs(t_cal),df)
print(f"P value:{p_value:.4f}")
if p_value<0.05:
    print("Reject null hypothesis")
else:
    print("Fail to reject null hypothesis")
```

```python
oi=[90,135,75]
# The sum of observed frequencies must equal the sum of expected frequencies.
# Assuming the proportions are 1/4, 1/2, 1/4, we scale them by the total sum of observed frequencies.
total_observed = sum(oi)
ei=[total_observed/4, total_observed/2, total_observed/4]
chisquare,p_value=stats.chisquare(oi,ei)
print(f"Chi square value:{chisquare:.4f}")
print(f"P value:{p_value:.4f}")
if p_value<0.05:
    print("Reject null hypothesis")
else:
    print("Fail to reject null hypothesis")
```
