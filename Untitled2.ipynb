```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs,make_moons
x,y=make_blobs(n_samples=200,n_features=2,centers=2,cluster_std=5,random_state=123)
plt.scatter(x[:,0],x[:,1],c=y)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Customer Data')
plt.show()
```

```python
x,y=make_moons(n_samples=500,noise=0.09,random_state=100)
plt.scatter(x[:,0],x[:,1],c=y)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Customer Data')
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
X=np.array([[1,2],[1,4],[1,0],[10,2],[10,4],[10,0],[12,0],[12,4],[14,2],[14,0]])
agglo=AgglomerativeClustering(n_clusters=2)
labels=agglo.fit_predict(X)
print(labels)
plt.scatter(X[:,0],X[:,1],c=labels)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Agglomerative Hierarchical Clustering')
plt.show()
```

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import dendrogram,linkage,fcluster
import pandas as pd
iris=load_iris()
X=iris.data
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
z=linkage(X_scaled,method='ward')
plt.figure(figsize=(12,6))
dendrogram(z)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
```

```python
iris = load_iris()
X = iris.data
feature_names = iris.feature_names

# Fit Hierarchical Clustering using complete linkage for 3 clusters
model = AgglomerativeClustering(n_clusters=3, linkage="complete")
labels = model.fit_predict(X)

# Create scatter plot (using Sepal Length vs Sepal Width)
plt.figure(figsize=(8, 5))
scatter = plt.scatter(
    X[:, 0], X[:, 1], c=labels, cmap="viridis", s=60, edgecolors="k"
)

plt.title("Iris Clustering: Agglomerative (Complete Linkage)")
plt.xlabel(feature_names[0])
plt.ylabel(feature_names[1])
plt.colorbar(scatter, label="Cluster ID")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
customer_data=pd.read_csv('customer_segmentation_hierarchical_clustering.csv')
customer_data.head()
```

```python
X=customer_data[['Annual_Income_k$','Spending_Score']]
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
agglo=AgglomerativeClustering(n_clusters=5)
labels=agglo.fit_predict(X_scaled)
print(labels)
plt.scatter(X.values[:,0],X.values[:,1],c=labels)
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.title('Customer Data')
plt.show()
```

```python
fig,ax=plt.subplots(1,3,figsize=(12,6))
z1=linkage(X_scaled,method='ward')
dendrogram(z1,ax=ax[0])
ax[0].set_xlabel('Annual Income')
ax[0].set_ylabel('Spending Score')
ax[0].set_title('Customer Data(Ward Linkage)')
z2=linkage(X_scaled,method='complete')
dendrogram(z2,ax=ax[1])
ax[1].set_xlabel('Annual Income')
ax[1].set_ylabel('Spending Score')
ax[1].set_title('Customer Data(Complete Linkage)')
z3=linkage(X_scaled,method='average')
dendrogram(z3,ax=ax[2])
ax[2].set_xlabel('Annual Income')
ax[2].set_ylabel('Spending Score')
ax[2].set_title('Customer Data(Average Linkage)')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
df = pd.DataFrame({
    'Annual Income (k$)': [15, 16, 17, 40, 42, 45, 70, 75, 80],
    'Spending Score (1-100)': [39, 81, 6, 77, 40, 42, 17, 88, 79],
    'Cluster': [0, 1, 2, 0, 1, 2, 0, 1, 2]
})
cluster_names={
    0: 'Mid value Customer',
    1: 'High value Customer',
    2: 'Low value Customer'
}
df['Cluster']=df['Cluster'].map(cluster_names)
# Calculate cluster-wise averages
cluster_avg = df.groupby('Cluster')[['Annual Income (k$)', 'Spending Score (1-100)']].mean().reset_index()

# Melt for seaborn grouped barplot
df_melted = cluster_avg.melt(id_vars='Cluster', var_name='Metric', value_name='Average Value')

plt.figure(figsize=(8, 5))
sns.barplot(data=df_melted, x='Cluster', y='Average Value', hue='Metric', palette='viridis')
plt.title('Average Income and Spending Score by Cluster')
plt.xlabel('Cluster')
plt.ylabel('Average Value')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
from sklearn.cluster import DBSCAN
import numpy as np
import matplotlib.pyplot as plt
X=np.array([[1,1],[1,2],[2,1],[2,2],[3,2],[8,8]])
dbscan=DBSCAN(eps=1.5,min_samples=3)
labels=dbscan.fit_predict(X)
print("Cluster Labels:")
print(labels)
print("\nCore Point Indices:")
print(dbscan.core_sample_indices_)
n_clusters=len(set(labels))-(1 if -1 in labels else 0)
print("\nNumber of Clusters:",n_clusters)
n_noise=list(labels).count(-1)
print("\nNumber of Noise Points:",n_noise)
plt.scatter(X[:,0],X[:,1],c=labels)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('DBSCAN Clustering')
plt.show()
```
