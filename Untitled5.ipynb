```python
import numpy as np
X=np.array([[2,1],[3,2],[4,3],[5,4]])
Xc=X-np.mean(X,axis=0)
c=np.cov(Xc,rowvar=False)
values,vectors=np.linalg.eig(c)
idx=np.argsort(values)[::-1]
values=values[idx]
vectors=vectors[:,idx]
PC1=vectors[:,0]
z=Xc @ PC1
print(values)
print(PC1)
print(z)
```

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
# Load the Iris dataset
iris = load_iris()
data = iris.data
labels = iris.target
label_names = iris.target_names
# Convert to a DataFrame for better visualization
iris_df = pd.DataFrame(data, columns=iris.feature_names)
# Perform PCA to reduce dimensionality to 2
pca = PCA(n_components=2)
data_reduced = pca.fit_transform(data)
# Create a DataFrame for the reduced data
reduced_df = pd.DataFrame(data_reduced, columns=['Principal Component 1', 'Principal Component 2'])
reduced_df['Label'] = labels
# Plot the reduced data
plt.figure(figsize=(8, 6))
colors = ['r', 'g', 'b']
for i, label in enumerate(np.unique(labels)):
  plt.scatter(reduced_df[reduced_df['Label'] == label]['Principal Component 1'], reduced_df[reduced_df['Label'] == label]['Principal Component 2'],label=label_names[label],color=colors[i])
plt.title('PCA on Iris Dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1. Generate realistic synthetic customer data
np.random.seed(42)
n_customers = 500

data = pd.DataFrame({
    'Annual_Income_Lakh': np.random.normal(12, 4, n_customers),        # ₹ lakh
    'Monthly_Spending_Thousand': np.random.normal(45, 15, n_customers), # ₹ thousand
    'Account_Balance_Lakh': np.random.normal(8, 3, n_customers),        # ₹ lakh
    'Loan_Amount_Lakh': np.random.normal(15, 6, n_customers)           # ₹ lakh
})

# 2. Standardize features (Z-score Normalization)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(data)

# 3. Fit PCA and select top 2 components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Create a DataFrame with the transformed dimensions
df_pca = pd.DataFrame(
    X_pca,
    columns=['PC1_Financial_Capacity', 'PC2_Credit_Leverage']
)

# 4. Display Explained Variance Ratio
print("--- Variance Explained ---")
for i, ratio in enumerate(pca.explained_variance_ratio_):
    print(f"PC{i+1}: {ratio:.2%}")
print(f"Total Explained Variance: {sum(pca.explained_variance_ratio_):.2%}\n")

# 5. Display Loadings Matrix (Interpretation of dimensions)
loadings = pd.DataFrame(
    pca.components_.T,
    index=data.columns,
    columns=['PC1', 'PC2']
)
print("--- Variable Loadings Matrix ---")
print(loadings.round(3))

# 6. Visualize 2D Customer Feature Space
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x='PC1_Financial_Capacity',
    y='PC2_Credit_Leverage',
    data=df_pca,
    alpha=0.7
)
plt.axhline(0, color='grey', linestyle='--', linewidth=0.8)
plt.axvline(0, color='grey', linestyle='--', linewidth=0.8)
plt.title('Customer Financial Segmentation Space (PCA Projection)')
plt.xlabel('PC1: Financial Capacity (Wealth & Spending)')
plt.ylabel('PC2: Credit Leverage (Debt Burden)')
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
import numpy as np
# Step 1: Create the dataset
# Rows = observations
# Columns = features
X = np.array([
    [2, 1],
    [4, 3],
    [6, 5]
], dtype=float)
print("Original Dataset:")
print(X)

# Step 2: Calculate the mean of each feature
mean = np.mean(X, axis=0)
print("\nMean of each feature:")
print(mean)
# Step 3: Center the data
X_centered = X - mean
print("\nCentered Data:")
print(X_centered)

# Step 4: Apply Singular Value Decomposition (SVD)
U, S, VT = np.linalg.svd(X_centered)
print("\nU Matrix:")
print(U)
print("\nSingular Values:")
print(S)
print("\nV Transpose Matrix:")
print(VT)


# Step 5: Principal Components
V = VT.T
print("\nPrincipal Component Directions (V):")
print(V)
# Step 6: Project the data onto Principal Components
Z = np.dot(X_centered, V)
print("\nPCA Transformed Data:")
print(Z)
# Step 7: Calculate Explained Variance
explained_variance = (S ** 2) / (X.shape[0] - 1)
print("\nExplained Variance:")
print(explained_variance)
# Step 8: Calculate Explained Variance Ratio
explained_variance_ratio = (S ** 2) / np.sum(S ** 2)
print("\nExplained Variance Ratio:")
print(explained_variance_ratio)
# Step 9: Keep only the first Principal Component
k = 1
V_k = V[:, :k]
Z_reduced = np.dot(X_centered, V_k)
print("\nReduced Data using First Principal Component:")
print(Z_reduced)
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
# -------------------------------------------------
# STEP 1: Create Customer Dataset
# -------------------------------------------------
data = {
    "Customer": ["C1", "C2", "C3", "C4", "C5", "C6"],
        "Annual_Income": [25, 30, 35, 45, 50, 60],
        "Spending_Score": [20, 35, 40, 55, 65, 80],
        "Purchases_Per_Month": [2, 3, 4, 6, 7, 9],
        "Average_Purchase": [500, 700, 800, 1200, 1500, 2000]
}
df = pd.DataFrame(data)
print("ORIGINAL DATASET")
print(df)
# -------------------------------------------------
# STEP 2: Select Numerical Features
# -------------------------------------------------
X = df[[
    "Annual_Income",
    "Spending_Score",
    "Purchases_Per_Month",
    "Average_Purchase"
]]

print("\nFEATURE MATRIX")
print(X)
# -------------------------------------------------
# STEP 3: Standardize the Dataset
# -------------------------------------------------
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print("\nSTANDARDIZED DATA")
print(X_scaled)
# -------------------------------------------------
# STEP 4: Apply SVD
# -------------------------------------------------
U, S, VT = np.linalg.svd(X_scaled, full_matrices=False)
print("\nU MATRIX")
print(U)
print("\nSINGULAR VALUES")
print(S)
print("\nV TRANSPOSE MATRIX")
print(VT)
# -------------------------------------------------
# STEP 5: Calculate Principal Components
# -------------------------------------------------
V = VT.T
print("\nPRINCIPAL COMPONENT DIRECTIONS")
print(V)
# -------------------------------------------------
# STEP 6: Calculate Explained Variance
# -------------------------------------------------
explained_variance = (S ** 2) / (len(X_scaled) - 1)
explained_variance_ratio = (
    explained_variance / np.sum(explained_variance)
)
print("\nEXPLAINED VARIANCE")
print(explained_variance)
print("\nEXPLAINED VARIANCE RATIO")
print(explained_variance_ratio)
# -------------------------------------------------
# STEP 7: Select First 2 Principal Components
# -------------------------------------------------
k = 2
V_k = V[:, :k]

# -------------------------------------------------
# STEP 8: Transform the Dataset
# ------------------------------------------------
X_pca = np.dot(X_scaled, V_k)
print("\nDATA AFTER PCA")
print(X_pca)
# -------------------------------------------------
# STEP 9: Create Final PCA Dataset
# -------------------------------------------------
pca_df = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)
pca_df["Customer"] = df["Customer"]
print("\nFINAL REDUCED DATASET")
print(pca_df)
```
