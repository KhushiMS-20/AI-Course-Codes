```python
import numpy as np
A=np.array([[4,7],
            [2,6]])
det_A=np.linalg.det(A)
if det_A!=0:
  A_inv=np.linalg.inv(A)
  print("Inverse of A:\n",A_inv)
else:print("A is singular and inverse does not exist.")
```

```python
import numpy as np
A=np.array([[2,3],
            [3,-6]])
eigenvalues,eigenvectors=np.linalg.eig(A)
print("Eigenvalues:\n",eigenvalues)
print("Eigenvectors:\n",eigenvectors)
```

```python
import sympy as sp
x=sp.symbols('x')
y=sp.symbols('y')
f=x**2+y**2-5
dfx=sp.diff(f,x)
dfy=sp.diff(f,y)
df=-dfx/dfy
print("Function f(x):",f)
print("Derivative of f(x):",sp.simplify(df))
```

```python
f=sp.sin(x)+x**3
df1=sp.diff(f,x)
df2=sp.diff(df1,x)
print("Function f(x):",f)
print("First derivative of f(x):",df1)
print("Second derivative of f(x):",df2)
```

```python
f=x**2*y+x*y**2+sp.sin(x*y)
dfx=sp.diff(f,x)
dfy=sp.diff(f,y)
print("Function f(x,y):",f)
print("Partial derivative with respect to x:",dfx)
print("Partial derivative with respect to y:",dfy)
gradient=sp.Matrix([dfx,dfy])
print("Gradient of f(x,y):",gradient)
```
